```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df=pd.read_csv(r"C:\Users\elmak\Downloads\archive (3)\advertising.csv")

df = df.drop(columns=['Ad Topic Line', 'City', 'Country', 'Timestamp', 'Male'])

print(df.columns.tolist())

df.head()
```

```python
scaler = StandardScaler()

scaler.fit(df)

scaled_data = scaler.transform(df)

scaled_data
```

```python
pca = PCA(n_components=2)

pca.fit(scaled_data)

scaled_pca = pca.transform(scaled_data)

scaled_data.shape
```

```python
scaled_pca.shape
```

```python
plt.figure(figsize=(10,8))

plt.scatter(scaled_pca[:, 0],scaled_pca[:, 1],c=df['Clicked on Ad'])

plt.xlabel('First Principal Component')

plt.ylabel('Second Principal Component')
```

```python
plt.figure(figsize=(10,8))

plt.scatter(df['Age'], df['Daily Internet Usage'])

plt.xlabel('Age')
plt.ylabel('Daily Internet Usage')
plt.show()
```
