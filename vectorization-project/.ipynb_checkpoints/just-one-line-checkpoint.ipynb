This model is trained on the vector data.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from PIL import Image, ImageDraw
```

```python
print(tf.__version__)
```

Create Synthetic data

```python
# one black line on white background

IMG_SIZE = 64

def create_line_image():
    # Random line coordinates
    x1, y1 = np.random.randint(0, IMG_SIZE, size=2)
    x2, y2 = np.random.randint(0, IMG_SIZE, size=2)

    # Create white image
    image = Image.new("L", (IMG_SIZE, IMG_SIZE), color=255)
    draw = ImageDraw.Draw(image)

    # Draw black line
    draw.line((x1, y1, x2, y2), fill=0, width=1)

    # Convert to numpy array
    image_array = np.array(image).astype("float32") / 255.0

    # CNN expects shape: height, width, channels
    image_array = np.expand_dims(image_array, axis=-1)

    # Normalise vector coordinates between 0 and 1
    label = np.array([
        x1 / IMG_SIZE,
        y1 / IMG_SIZE,
        x2 / IMG_SIZE,
        y2 / IMG_SIZE
    ], dtype="float32")

    return image_array, label
```

```python
#data set generation

def create_dataset(n_samples):
    images = []
    labels = []

    for _ in range(n_samples):
        image, label = create_line_image()
        images.append(image)
        labels.append(label)

    return np.array(images), np.array(labels)

X_train, y_train = create_dataset(5000)
X_test, y_test = create_dataset(1000)

print(X_train.shape)
print(y_train.shape)
```

```python
#display one

index = 99

plt.imshow(X_train[index].squeeze(), cmap="gray")
plt.title(f"Vector label: {y_train[index]}")
plt.axis("off")
plt.show()
```

```python
#Cnn

model = models.Sequential([
    layers.Input(shape=(64, 64, 1)),

    layers.Conv2D(16, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),

    # 4 values: x1, y1, x2, y2
    layers.Dense(4, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
#Train

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=20,
    batch_size=32
)
```

```python
#plot loss

plt.plot(history.history["loss"], label="Training loss")
plt.plot(history.history["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
#plot historyplt.show()
```

```python
#plot history

plt.plot(history.history["mae"], label="Training MAE")
plt.plot(history.history["val_mae"], label="Validation MAE")
plt.xlabel("Epoch")
plt.ylabel("Mean Absolute Error")
plt.legend()
plt.show()
```

```python
#prediction

sample = X_test[0:1]
true_vector = y_test[0]

predicted_vector = model.predict(sample)[0]

print("True vector:", true_vector)
print("Predicted vector:", predicted_vector)
```

```python
# convert back to pixels

def denormalise_vector(vector):
    return (vector * IMG_SIZE).astype(int)

print("True pixels:", denormalise_vector(true_vector))
print("Predicted pixels:", denormalise_vector(predicted_vector))
```

```python
#raster prediction

def rasterise_line(vector):
    x1, y1, x2, y2 = denormalise_vector(vector)

    image = Image.new("L", (IMG_SIZE, IMG_SIZE), color=255)
    draw = ImageDraw.Draw(image)
    draw.line((x1, y1, x2, y2), fill=0, width=1)

    image_array = np.array(image).astype("float32") / 255.0
    return image_array
```

```python
#display prediction verses raster

predicted_raster = rasterise_line(predicted_vector)

plt.imshow(sample.squeeze(), cmap="gray")
plt.title("Input raster")
plt.axis("off")
plt.show()

plt.imshow(predicted_raster, cmap="gray")
plt.title("Rasterised predicted vector")
plt.axis("off")
plt.show()
```
