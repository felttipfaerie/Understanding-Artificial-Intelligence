# One-line CNN with differentiable rasterisation

This version changes the training target from vector-coordinate loss to raster-image loss.

The CNN still predicts `[x1, y1, x2, y2]`, but those predicted coordinates are passed through a TensorFlow rasteriser layer. The model is then trained by comparing the rasterised prediction with the original 64×64 image.

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from PIL import Image, ImageDraw

print(tf.__version__)
````

Code was generated with ChatGPT as the coding agent and then edited

````python
# One black line on white background

IMG_SIZE = 64

def create_line_image():
    # Random line coordinates
    x1, y1 = np.random.randint(0, IMG_SIZE, size=2)
    x2, y2 = np.random.randint(0, IMG_SIZE, size=2)

    # Create white image
    image = Image.new("L", (IMG_SIZE, IMG_SIZE), color=255)
    draw = ImageDraw.Draw(image)

    # Draw black line
    draw.line((x1, y1, x2, y2), fill=0, width=1)

    # Convert to numpy array: white = 1, black = 0
    image_array = np.array(image).astype("float32") / 255.0

    # CNN expects shape: height, width, channels
    image_array = np.expand_dims(image_array, axis=-1)

    # Normalise vector coordinates between 0 and 1
    vector_label = np.array([
        x1 / (IMG_SIZE - 1),
        y1 / (IMG_SIZE - 1),
        x2 / (IMG_SIZE - 1),
        y2 / (IMG_SIZE - 1)
    ], dtype="float32")

    return image_array, vector_label
````

````python
# Dataset generation

def create_dataset(n_samples):
    images = []
    vector_labels = []

    for _ in range(n_samples):
        image, vector_label = create_line_image()
        images.append(image)
        vector_labels.append(vector_label)

    return np.array(images), np.array(vector_labels)

X_train, y_train_vectors = create_dataset(5000)
X_test, y_test_vectors = create_dataset(1000)

print("Images:", X_train.shape)
print("Vector labels:", y_train_vectors.shape)
````

````python
# Display one training example

index = 99

plt.imshow(X_train[index].squeeze(), cmap="gray")
plt.title(f"True vector label: {y_train_vectors[index]}")
plt.axis("off")
plt.show()
````

## Differentiable rasteriser

This layer replaces the old PIL `rasterise_line()` function during training.

PIL can draw a line, but TensorFlow cannot backpropagate through PIL. This layer uses TensorFlow maths only. It calculates the distance from every pixel to the predicted line segment, then creates a soft black line where pixels close to the line become dark.

````python
class DifferentiableLineRasteriser(layers.Layer):
    def __init__(self, img_size=64, sigma=0.012, **kwargs):
        super().__init__(**kwargs)
        self.img_size = img_size
        self.sigma = sigma

        # Pixel centres, normalised between 0 and 1
        coords = tf.linspace(0.0, 1.0, img_size)
        yy, xx = tf.meshgrid(coords, coords, indexing="ij")

        # Shape: (1, height, width, 2)
        pixel_grid = tf.stack([xx, yy], axis=-1)
        self.pixel_grid = tf.reshape(pixel_grid, (1, img_size, img_size, 2))

    def call(self, vectors):
        # vectors shape: (batch, 4)
        # a = start point, b = end point
        a = vectors[:, 0:2]  # x1, y1
        b = vectors[:, 2:4]  # x2, y2

        # Reshape for broadcasting over image grid
        a = tf.reshape(a, (-1, 1, 1, 2))
        b = tf.reshape(b, (-1, 1, 1, 2))

        p = self.pixel_grid
        ab = b - a
        ap = p - a

        # Project each pixel onto the line segment
        ab_squared = tf.reduce_sum(ab * ab, axis=-1, keepdims=True) + 1e-8
        t = tf.reduce_sum(ap * ab, axis=-1, keepdims=True) / ab_squared
        t = tf.clip_by_value(t, 0.0, 1.0)

        closest = a + t * ab

        # Distance from each pixel to the nearest point on the line segment
        distance = tf.sqrt(
            tf.reduce_sum((p - closest) ** 2, axis=-1, keepdims=True) + 1e-8
        )

        # Soft black line on white background
        # line_strength close to 1 near the line, close to 0 far away
        line_strength = tf.exp(-(distance ** 2) / (2.0 * self.sigma ** 2))

        # Match your image convention: white = 1, black = 0
        raster = 1.0 - line_strength

        return raster
````

## Build two connected models

`vector_model` lets you inspect the predicted coordinates.

`training_model` adds the differentiable rasteriser and is the model you train.

````python
# CNN predicts vector coordinates

inputs = layers.Input(shape=(64, 64, 1))

x = layers.Conv2D(16, (3, 3), activation="relu", padding="same")(inputs)
x = layers.MaxPooling2D((2, 2))(x)

x = layers.Conv2D(32, (3, 3), activation="relu", padding="same")(x)
x = layers.MaxPooling2D((2, 2))(x)

x = layers.Conv2D(64, (3, 3), activation="relu", padding="same")(x)
x = layers.MaxPooling2D((2, 2))(x)

x = layers.Flatten()(x)

x = layers.Dense(128, activation="relu")(x)
x = layers.Dense(64, activation="relu")(x)

# Four normalised coordinate values: x1, y1, x2, y2
vector_output = layers.Dense(4, activation="sigmoid", name="vector_output")(x)

vector_model = models.Sequential([
    layers.Input(shape=(64, 64, 1)),

    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
    layers.Flatten(),

    layers.Dense(256, activation="relu"),
    layers.Dense(128, activation="relu"),

    # x1, y1, x2, y2
    layers.Dense(4, activation="sigmoid")
])

# Differentiable raster output
raster_output = DifferentiableLineRasteriser(
    img_size=IMG_SIZE,
    sigma=0.012,
    name="differentiable_rasteriser"
)(vector_output)

training_model = models.Model(inputs=inputs, outputs=raster_output, name="raster_training_model")

training_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

training_model.summary()
````

## Train using the raster image as the target

This is the biggest practical change.

Old version:

```python
model.fit(X_train, y_train)
```

New Option B version:

```python
training_model.fit(X_train, X_train)
```

The input image is also the target image because the model predicts a vector, rasterises it, and compares that raster with the original raster.

````python
history = training_model.fit(
    X_train,
    X_train,
    validation_data=(X_test, X_test),
    epochs=20,
    batch_size=32
)
````

````python
# Plot raster reconstruction loss

plt.plot(history.history["loss"], label="Training loss")
plt.plot(history.history["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Raster MSE")
plt.legend()
plt.show()
````

The drop off shows that the model is not training well. It's being "lazy" and just predicting a dot and mostly white

＞﹏＜

````python
# Plot raster reconstruction error

plt.plot(history.history["mae"], label="Training MAE")
plt.plot(history.history["val_mae"], label="Validation MAE")
plt.xlabel("Epoch")
plt.ylabel("Raster MAE")
plt.legend()
plt.show()
````

## Inspect the predicted vector and raster output

````python
sample = X_test[0:1]

true_vector = y_test_vectors[0]
predicted_vector = vector_model.predict(sample)[0]
predicted_raster = training_model.predict(sample)[0]

print("True vector:", true_vector)
print("Predicted vector:", predicted_vector)
````

````python
def denormalise_vector(vector):
    return np.round(vector * (IMG_SIZE - 1)).astype(int)

print("True pixels:", denormalise_vector(true_vector))
print("Predicted pixels:", denormalise_vector(predicted_vector))
````

````python
plt.imshow(sample.squeeze(), cmap="gray")
plt.title("Input / target raster")
plt.axis("off")
plt.show()

plt.imshow(predicted_raster.squeeze(), cmap="gray")
plt.title("Differentiably rasterised predicted vector")
plt.axis("off")
plt.show()
````

## Optional: hard rasterise for final display/export

This is only for display or SVG export after training. Do not use this PIL version inside the training loss.

````python
def hard_rasterise_line(vector):
    x1, y1, x2, y2 = denormalise_vector(vector)

    image = Image.new("L", (IMG_SIZE, IMG_SIZE), color=255)
    draw = ImageDraw.Draw(image)
    draw.line((x1, y1, x2, y2), fill=0, width=1)

    image_array = np.array(image).astype("float32") / 255.0
    return image_array

hard_predicted_raster = hard_rasterise_line(predicted_vector)

plt.imshow(hard_predicted_raster, cmap="gray")
plt.title("Hard rasterised prediction for final display")
plt.axis("off")
plt.show()
````

````python
def line_to_svg(vector, width=64, height=64):
    x1, y1, x2, y2 = denormalise_vector(vector)

    svg = f'''<svg width="{width}" height="{height}" xmlns="http://www.w3.org/2000/svg">
    <line x1="{x1}" y1="{y1}" x2="{x2}" y2="{y2}"
          stroke="black" stroke-width="1"/>
</svg>'''
    return svg

print(line_to_svg(predicted_vector))
````
